# Завдання 1

```python
import matplotlib.pyplot as plt

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
plan = [100, 115, 125, 130, 145, 150, 165, 160, 175, 180, 190, 210]
fact = [95, 120, 120, 135, 140, 155, 160, 165, 170, 185, 195, 205]

plt.figure(figsize=(10, 5))
plt.plot(months, plan, label="План", marker="o")
plt.plot(months, fact, label="Факт", marker="s")
plt.xlabel("Місяць")
plt.ylabel("Продажі")
plt.title("Порівняння планових та фактичних продажів")
plt.legend()
plt.show()
```

# Завдання 2

```python
import numpy as np
import matplotlib.pyplot as plt

ages = np.random.randint(18, 70, size=100)
mean_age = np.mean(ages)

plt.figure(figsize=(8, 5))
plt.hist(ages, bins=10, edgecolor="black")
plt.axvline(mean_age, color="red", linestyle="--", linewidth=2, label=f"Середній вік: {mean_age:.1f}")
plt.xlabel("Вік")
plt.ylabel("Кількість осіб")
plt.title("Розподіл віку серед 100 осіб")
plt.legend()
plt.show()
```

# Завдання 3

```python
import numpy as np
import matplotlib.pyplot as plt

group_1 = np.random.randint(50, 100, size=25)
group_2 = np.random.randint(60, 100, size=25)
group_3 = np.random.randint(40, 95, size=25)

plt.figure(figsize=(8, 6))
plt.boxplot([group_1, group_2, group_3], labels=["Група 1", "Група 2", "Група 3"])
plt.xlabel("Групи студентів")
plt.ylabel("Оцінки за іспит")
plt.title("Розкид оцінок за іспит між трьома групами")
plt.show()
```
